1.Even or Odd Function

```python
def is_even(n):
    if n % 2 == 0:
        return "Even"
    else:
        return "Odd"

num = int(input("Enter a number: "))
print(is_even(num))
```

2. Largest of Two Numbers

```python
def largest(a, b):
    if a > b:
        return a
    else:
        return b

print(largest(10, 20))
```

3. Print Numbers from 1 to n

```python
def print_numbers(n):
    for i in range(1, n + 1):
        print(i)

print_numbers(5)
```

4. Sum of Numbers from 1 to n

```python
def sum_n(n):
    total = 0
    for i in range(1, n + 1):
        total = total + i
    return total

print(sum_n(5))
```

5. Multiplication Table

```python
def table(n):
    for i in range(1, 11):
        print(n, "*", i, "=", n * i)

table(5)
```

6. Count Even Numbers in a List

```python
def count_even(lst):
    count = 0
    for i in lst:
        if i % 2 == 0:
            count = count + 1
    return count

print(count_even([1, 2, 3, 4, 6]))
```

7. Find Maximum in a List (Without max())

```python
def find_max(lst):
    largest = lst[0]

    for i in lst:
        if i > largest:
            largest = i

    return largest

print(find_max([10, 25, 5, 40, 15]))
```

11. FizzBuzz

```python
def fizzbuzz():
    for i in range(1, 101):
        if i % 3 == 0 and i % 5 == 0:
            print("FizzBuzz")
        elif i % 3 == 0:
            print("Fizz")
        elif i % 5 == 0:
            print("Buzz")
        else:
            print(i)

fizzbuzz()
```

12. Palindrome Checker

```python
def palindrome(s):
    if s == s[::-1]:
        return True
    else:
        return False

print(palindrome("madam"))
```

13. Count Vowels

```python
def count_vowels(s):
    count = 0
    for ch in s.lower():
        if ch in "aeiou":
            count = count + 1
    return count

print(count_vowels("Hello World"))
```

14. Second Largest Number (Without sort())

```python
def second_largest(lst):
    largest = lst[0]
    second = None

    for i in lst:
        if i > largest:
            second = largest
            largest = i
        elif i < largest:
            if second is None or i > second:
                second = i

    return second

print(second_largest([10, 20, 5, 30, 25]))
```

15. Guessing Game

```python
import random

def guessing_game():
    num = random.randint(1, 10)

    while True:
        guess = int(input("Guess a number between 1 and 10: "))

        if guess == num:
            print("Correct guess!")
            break
        else:
            print("Try again!")
    print("The number was:", num)        

guessing_game()
```

16. Grade Function

```python
def grade(marks):
    if marks >= 90:
        return "A"
    elif marks >= 80:
        return "B"
    elif marks >= 70:
        return "C"
    elif marks >= 60:
        return "D"
    else:
        return "F"

print(grade(85))
```
